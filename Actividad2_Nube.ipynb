```python
ventas_i, ventas_f = 200000, 260000
costos_i, costos_f = 120000, 150000

pct_ventas = (ventas_f - ventas_i) / ventas_i * 100
pct_costos = (costos_f - costos_i) / costos_i * 100

utilidad_i = ventas_i - costos_i
utilidad_f = ventas_f - costos_f
pct_utilidad = (utilidad_f - utilidad_i) / utilidad_i * 100

print(f"a) Aumento en ventas: {pct_ventas}%")
print(f"b) Aumento en costos: {pct_costos}%")
print(f"c) Utilidad inicial: {utilidad_i}, final: {utilidad_f}, aumento: {pct_utilidad}%")
```
